```python
# ==========================================
#   SISTEMA IA - CONTROL DE POLICÍAS
#   Programación Orientada a Objetos
# ==========================================
class Policia:
    # Constructor
    def __init__(self, placa, patrulla, sector, anios, rango, curso):
        self.placa = placa
        self.patrulla = patrulla
        self.sector = sector
        self.anios = anios
        self.rango = rango
        self.curso = curso
    # Mostrar información
    def mostrar_datos(self):
        print(f"{self.placa:<15}{self.patrulla:<15}{self.sector:<15}{self.anios:<10}{self.rango:<15}{self.curso:<25}")
    # IA simple para evaluación
    def evaluar_policia(self):
        if self.anios >= 15:
            experiencia = "EXPERTO"
        elif self.anios >= 8:
            experiencia = "AVANZADO"
        else:
            experiencia = "NOVATO"
        print("\n===== ANALISIS IA =====")
        print("Placa:", self.placa)
        print("Rango:", self.rango)
        print("Experiencia detectada:", experiencia)
        if self.curso.lower() == "tactico":
            print("IA: Especialista en operaciones tácticas.")
        elif self.curso.lower() == "primeros auxilios":
            print("IA: Capacitado para emergencias médicas.")
        elif self.curso.lower() == "investigacion":
            print("IA: Perfil apto para investigación.")
        else:
            print("IA: Curso general registrado.")
# ==========================================
# LISTA DE POLICÍAS
# ==========================================
lista_policias = []
# ==========================================
# MENÚ PRINCIPAL
# ==========================================
while True:
    print("\n")
    print("====================================")
    print("    SISTEMA IA DE POLICÍAS")
    print("====================================")
    print("1. Registrar policía")
    print("2. Mostrar policías")
    print("3. Analizar policía")
    print("4. Salir")
    print("====================================")
    opcion = input("Seleccione una opción: ")
    # ======================================
    # REGISTRAR POLICÍA
    # ======================================
    if opcion == "1":
        print("\n--- REGISTRO DE POLICÍA ---")
        placa = input("Número de placa: ")
        patrulla = input("Patrulla asignada: ")
        sector = input("Sector de vigilancia: ")
        anios = int(input("Años de servicio: "))
        rango = input("Rango: ")
        curso = input("Curso de especialización: ")
        nuevo_policia = Policia(
            placa,
            patrulla,
            sector,
            anios,
            rango,
            curso
        )
        lista_policias.append(nuevo_policia)
        print("\nPolicía registrado correctamente.")
    # ======================================
    # MOSTRAR POLICÍAS
    # ======================================
    elif opcion == "2":
        print("\n================ LISTA DE POLICÍAS ================\n")
        if len(lista_policias) == 0:
            print("No hay policías registrados.")
        else:
            print(f"{'PLACA':<15}{'PATRULLA':<15}{'SECTOR':<15}{'AÑOS':<10}{'RANGO':<15}{'CURSO':<25}")
            print("-" * 95)
            for policia in lista_policias:
                policia.mostrar_datos()
    # ======================================
    # ANALIZAR POLICÍA
    # ======================================
    elif opcion == "3":
        if len(lista_policias) == 0:
            print("\nNo existen policías registrados.")
        else:
            buscar = input("\nIngrese número de placa: ")
            encontrado = False
            for policia in lista_policias:
                if policia.placa == buscar:
                    policia.evaluar_policia()
                    encontrado = True
            if encontrado == False:
                print("Policía no encontrado.")
    # ======================================
    # SALIR
    # ======================================
    elif opcion == "4":
        print("\nSaliendo del sistema IA...")
        print("Creado por Daniel Nieves NL=45")
        break
    # ======================================
    # OPCIÓN INVÁLIDA
    # ======================================
    else:
        print("\nOpción inválida.")
```
